# FraudShield AI — Exploratory Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

%matplotlib inline
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('../data/creditcard.csv')
print(f"Dataset Shape: {df.shape}")
df.info()
```

## Class Distribution

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

sns.countplot(data=df, x='Class', ax=ax1)
ax1.set_title('Class Distribution (Bar Chart)')
ax1.set_yscale('log')
ax1.set_ylabel('Count (Log Scale)')

class_counts = df['Class'].value_counts()
ax2.pie(class_counts, labels=['Legitimate (0)', 'Fraud (1)'], autopct='%1.3f%%', colors=['#2ecc71', '#e74c3c'])
ax2.set_title('Class Distribution (Pie Chart)')

plt.tight_layout()
plt.show()
```

## Amount Distribution by Class

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.boxplot(data=df, x='Class', y='Amount', ax=ax)
ax.set_yscale('log')
ax.set_title('Transaction Amount by Class (Log Scale)')
ax.set_xticklabels(['Legitimate (0)', 'Fraud (1)'])
plt.show()
```

## Fraud by Hour of Day

```python
df['hour_of_day'] = (df['Time'] % 86400) // 3600

plt.figure(figsize=(12, 6))
sns.histplot(data=df[df['Class'] == 1], x='hour_of_day', bins=24, kde=True, color='red')
plt.title('Fraudulent Transactions by Hour of Day')
plt.xlabel('Hour of Day (0-23)')
plt.ylabel('Count')
plt.xticks(range(0, 24))
plt.show()
```

## Correlation with Fraud

```python
plt.figure(figsize=(12, 10))
corr = df.drop(['Time', 'hour_of_day'], axis=1).corr()
sns.heatmap(corr[['Class']].sort_values(by='Class', ascending=False), 
            vmin=-1, vmax=1, annot=True, cmap='coolwarm')
plt.title('Feature Correlation with Class (Fraud)')
plt.show()
```

## Key Findings

1. **Extreme Class Imbalance**: Fraud accounts for only 0.172% of all transactions. This necessitates specific handling (e.g., SMOTE, class weights, or specific evaluation metrics like PR-AUC).
2. **Amount**: Fraudulent transactions tend to have smaller, more concentrated amounts, though there are outliers.
3. **Time**: Fraud appears to occur more evenly throughout the day, while legitimate transactions drop heavily during nighttime hours.
4. **Correlations**: Features like V14, V12, and V17 show strong negative correlations with fraud, while V11 and V4 show positive correlations. This indicates the PCA components successfully captured variance relevant to the classification task.
